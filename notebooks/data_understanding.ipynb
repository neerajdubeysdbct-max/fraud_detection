## data exploration on cash out and transfer

```python
# create a scoped dataset with only TRANSFER and CASH_OUT transactions
df_scope = df_parquet[df_parquet['type'].isin(['TRANSFER', 'CASH_OUT'])].copy()
df_scope.shape
df_scope['type'].value_counts()
```

```python
pd.crosstab(df_scope['type'], df_scope['isFraud'])
```

```python
# undersrand the fraud rate by transaction type in the scoped dataset
fraud_rate = (
    df_scope.groupby('type')['isFraud']
    .mean()
    .mul(100)
)

print(fraud_rate)
```

```python
#comparinf the amount of fraudulent and non-fraudulent transactions in the scoped dataset
df_scope.groupby('isFraud')['amount'].agg(
    ['count', 'mean', 'median', 'min', 'max']
)
```

```python
# comparing within the transaction types in the scoped dataset
df_scope.groupby(
    ['type', 'isFraud']
)['amount'].agg(
    ['count', 'mean', 'median', 'min', 'max']
)
```

```python
# create a new column to indicate if the account was emptied after the transaction
df_scope['account_emptied'] = (
    df_scope['newbalanceOrig'] == 0
).astype(int)
```

```python
pd.crosstab(
    df_scope['account_emptied'],
    df_scope['isFraud'],
    normalize='index'
) * 100
```

```python
# by transaction type and account emptied status
pd.crosstab(
    [df_scope['type'], df_scope['account_emptied']],
    df_scope['isFraud'],
    normalize='index'
) * 100
```

```python
# balance inconsistencies in the scoped dataset
df_scope['orig_balance_error'] = (
    df_scope['oldbalanceOrg']
    - df_scope['amount']
    - df_scope['newbalanceOrig']
)
```

```python
df_scope.groupby(
    ['type', 'isFraud']
)['orig_balance_error'].agg(
    ['mean', 'median', 'min', 'max']
)
```

```python
# checking the absolute error in the original balance after the transaction
df_scope['orig_balance_error_abs'] = (
    df_scope['orig_balance_error'].abs()
)
```

```python
df_scope.groupby(
    ['type', 'isFraud']
)['orig_balance_error_abs'].mean()
```

```python
df_scope['account_emptied'] = (
    df_scope['newbalanceOrig'] == 0
).astype(int)
```

```python
dest_transaction_count = (
    df_scope['nameDest']
    .value_counts()
)

df_scope['dest_transaction_count'] = (
    df_scope['nameDest']
    .map(dest_transaction_count)
)
```

```python
df_scope.groupby('isFraud')['dest_transaction_count'].agg(
    ['mean', 'median', 'min', 'max']
)
```

```python
df_scope.groupby(
    ['type', 'isFraud']
)['dest_transaction_count'].agg(
    ['mean', 'median']
)
```
